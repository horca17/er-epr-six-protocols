# Reproduce Experiment 1 — Mermin operator

Retrieves job `d7g38hma0v2s738abtf0` from IBM Quantum and recomputes |⟨M⟩| on the GHZ state (θ = 0).

```python
from qiskit_ibm_runtime import QiskitRuntimeService
import numpy as np

service = QiskitRuntimeService(channel="ibm_cloud")

def check(name, value, paper, tol):
    ok = abs(value - paper) <= tol
    print(f"{name:32s} recomputed {value:9.4f}   paper {paper:9.4f}   {'OK' if ok else 'DIFFERENT'}")
```

```python
job = service.job("d7g38hma0v2s738abtf0")
evs = np.abs(job.result()[0].data.evs.flatten())
thetas = np.linspace(0, np.pi/2, 11)
for t, m in zip(thetas, evs):
    print(f"theta/pi = {t/np.pi:.2f}   |<M>| = {m:.3f}")
```

```python
check("|<M>| at theta = 0", evs[0], 3.885, 0.0005)
```
