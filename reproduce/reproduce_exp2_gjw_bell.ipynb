# Reproduce Experiment 2 — GJW wormhole, Bell-pair TFD

Retrieves job `d7gft2jjne2c7392r44g` and recomputes the traversal fidelity F = (1 + ⟨X⟩)/2. The paper reports F clipped to [0, 1]; the raw value is printed too.

```python
from qiskit_ibm_runtime import QiskitRuntimeService
import numpy as np

service = QiskitRuntimeService(channel="ibm_cloud")

def check(name, value, paper, tol):
    ok = abs(value - paper) <= tol
    print(f"{name:32s} recomputed {value:9.4f}   paper {paper:9.4f}   {'OK' if ok else 'DIFFERENT'}")
```

```python
job = service.job("d7gft2jjne2c7392r44g")
evs = job.result()[0].data.evs.flatten()
F_raw = (1 + evs) / 2
F = np.clip(F_raw, 0, 1)
for g, e, f in zip(np.linspace(0, np.pi/2, 11), evs, F_raw):
    print(f"g/pi = {g/np.pi:.2f}   <X> = {e:7.4f}   F_raw = {f:.4f}")
```

```python
check("F(g=0)", F[0], 0.516, 0.0005)
check("F(g=pi/2), clipped", F[-1], 1.000, 0.0005)
print(f"F(g=pi/2) raw = {F_raw[-1]:.4f}")
```
